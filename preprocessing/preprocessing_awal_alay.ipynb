```python
import pandas as pd
import re
from tqdm import tqdm

print("Load Kamus Alay...")
df_kamus = pd.read_csv('../resources/kamus_alay_master.csv')
kamus_alay_gabungan = dict(zip(df_kamus['slang'], df_kamus['formal']))

def clean_text(text):
    text = str(text)
    
    words = text.split()
    normalized_words = [kamus_alay_gabungan.get(word, word) for word in words]
    text_normalized = ' '.join(normalized_words)

    text_final = re.sub(r'\s+', ' ', text_normalized).strip()
    
    return text_final

df = pd.read_csv('../resources/test_preprocessing_awal.csv')
print(f"Jumlah data mentah awal: {len(df)}")

tqdm.pandas(desc="Proses Cleaning berjalan")

df['Pesan_Clean'] = df['Pesan_Clean'].progress_apply(clean_text)

df = df[df['Pesan_Clean'] != '']
 
print("\nContoh Hasil Perbandingan (15 Teratas):")
print(df[['Pesan', 'Pesan_Clean']].head(15))

df.to_csv('../resources/test_preprocessing_alay.csv', index=False)
```
