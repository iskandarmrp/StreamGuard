```python
import pandas as pd

# Download stopwords
df_stopwords = pd.read_csv('https://raw.githubusercontent.com/datascienceid/stopwords-bahasa-indonesia/master/stopwords_id_satya.txt', header=None)
df_stopwords.to_csv('../resources/stopwords.csv', index=False)
```

```python
import pandas as pd

print("Mengunduh dan menggabungkan kamus...")

# Load Kamus 1 (Fendi Irfan)
df_kamus1 = pd.read_csv('https://raw.githubusercontent.com/fendiirfan/Kamus-Alay/main/Kamus-Alay.csv')
df_kamus1 = df_kamus1.rename(columns={'kataAlay': 'slang', 'kataBaik': 'formal'})

# Load Kamus 2 (Nasal Sabila)
df_kamus2 = pd.read_csv('https://raw.githubusercontent.com/nasalsabila/kamus-alay/master/colloquial-indonesian-lexicon.csv')
df_kamus2 = df_kamus2[['slang', 'formal']]

# Gabungkan kedua DataFrame
df_gabungan = pd.concat([df_kamus1, df_kamus2], ignore_index=True)

# Hapus duplikat
df_gabungan = df_gabungan.drop_duplicates(subset=['slang'], keep='first')

df_gabungan.to_csv('../resources/kamus_alay_master.csv', index=False)

print(f"Berhasil! Total kosa kata saat ini: {len(df_gabungan)} kata.\n")
```
