```python
import pandas as pd
from tqdm import tqdm
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

print("Mempersiapkan Stemmer PySastrawi...")
factory = StemmerFactory()
stemmer = factory.create_stemmer()

print("Load Daftar Stopwords...")
df_stopwords = pd.read_csv('../resources/stopwords.csv', header=None)
kumpulan_stopwords = set(df_stopwords[0]) 
custom_stopwords = {'sih', 'nya', 'mah', 'tuh'} 
kumpulan_stopwords.update(custom_stopwords)

def clean_text(text):
    text = str(text)

    text_stemmed = stemmer.stem(text)
    
    words = text_stemmed.split()
    filtered_words = [word for word in words if word not in kumpulan_stopwords]
    
    text_final = ' '.join(filtered_words)
    
    return text_final


df = pd.read_csv('../resources/test_preprocessing_awal.csv')
print(f"Jumlah data mentah awal: {len(df)}")

tqdm.pandas(desc="Proses Cleaning berjalan")

df['Pesan_Clean'] = df['Pesan_Clean'].progress_apply(clean_text)

df = df[df['Pesan_Clean'] != '']

print(f"\nJumlah data akhir siap pakai: {len(df)}")

print("\nContoh Hasil Perbandingan (15 Teratas):")
print(df[['Pesan', 'Pesan_Clean']].head(15))

df.to_csv('../resources/test_preprocessing_awal_shallow.csv', index=False)
```
