```python
import pandas as pd
import re
import unicodedata

# Load data
df = pd.read_csv('../scraping/data_scraping/data_chat_youtube_kobo_1_4_2022.csv')
print(f"Jumlah data mentah awal: {len(df)}")

# Cleaning
def clean_text(text):
    text = str(text)
    
    # Menghapus Emote bawaan YouTube (format :nama_emote:)
    text = re.sub(r':[^:\s]+:', ' ', text)

    # Normalisasi karakter Unicode (mengubah font unik/tebal/miring menjadi huruf standar)
    text = unicodedata.normalize('NFKD', text)
    
    # Menghapus karakter non-ASCII (Emoji standar, dll)
    text = text.encode('ascii', 'ignore').decode('ascii')
    
    # Menghapus URL/Link
    # text = re.sub(r'http\S+|www\.\S+', '', text)
    
    # Lowercase
    text = text.lower()
    
    # Menghapus angka dan tanda baca
    # text = re.sub(r'[^a-z\s]', ' ', text)
    
    # Menghapus ekstra spasi ganda dan spasi di awal/akhir kalimat
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

df['Pesan_Clean'] = df['Pesan'].apply(clean_text)
print(f"Jumlah pesan: {len(df)}")

# Hapus baris yang pesannya menjadi kosong
df = df[df['Pesan_Clean'] != '']
print(f"Jumlah data setelah emote & pesan kosong dihapus: {len(df)}")

# Hapus Duplikat
df = df.drop_duplicates(subset=['Pesan_Clean'])
print(f"Jumlah data akhir siap pakai: {len(df)}")

# 10000 dari belakang (data terbaru)
df = df.tail(10000)
print(f"Jumlah data akhir siap pakai (Max 10.000): {len(df)}")

# Simpan file
df.to_csv('preprocessing_output/data_chat_youtube_kobo_1_4_2022_cleaned.csv', index=False)
```
