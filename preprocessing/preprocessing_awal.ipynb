```python
import pandas as pd
import re
import unicodedata
from tqdm import tqdm

# Preprocessing awal
def clean_text(text):
    text = str(text)
    
    # Menghapus Emote bawaan YouTube (format :nama_emote:)
    text = re.sub(r':[^:\s]+:', ' ', text)

    # Normalisasi Unicode (NFKD) & Hapus karakter non-ASCII (Emoji)
    text = unicodedata.normalize('NFKD', text).encode('ascii', 'ignore').decode('ascii')
    
    # Menghapus URL/Link
    text = re.sub(r'http\S+|www\.\S+', '', text)

    # Lowercase
    text = text.lower()
    
    # Menghapus Karakter Berulang
    text = re.sub(r'(.)\1{2,}', r'\1', text)
    
    # Menghapus tanda baca
    text = re.sub(r'[^a-z0-9\s]', ' ', text)

    # Menghapus ekstra spasi ganda dan spasi di awal/akhir kalimat
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

df = pd.read_csv('../resources/test.csv')
print(f"Jumlah data mentah awal: {len(df)}")

tqdm.pandas(desc="Proses Cleaning berjalan")

df['Pesan_Clean'] = df['Pesan'].progress_apply(clean_text)

df = df[df['Pesan_Clean'] != '']
print(f"\nJumlah data akhir siap pakai: {len(df)}")

print("\nContoh Hasil Perbandingan (15 Teratas):")
print(df[['Pesan', 'Pesan_Clean']].head(15))

df.to_csv('../resources/test_preprocessing_awal.csv', index=False)
```
