```python
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import CountVectorizer

# Load data
hf_token = "..."

ds = load_dataset(
    "manueltonneau/indonesian-hate-speech-superset", 
    token=hf_token
)

df = ds['train'].to_pandas()

print(f"\nTotal data: {len(df)} baris")
print("\nInfo Dataset:")
print(df.info())

print("\n5 Data Pertama:")
display(df.head())

# Distribusi Label
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='labels', palette='Set2')
plt.title('Distribusi Label (0: Non-Hate Speech, 1: Hate Speech)')
plt.xlabel('Label')
plt.ylabel('Jumlah Teks')
for p in plt.gca().patches:
    plt.gca().annotate(f'{p.get_height()}', (p.get_x() + 0.3, p.get_height() + 500))
plt.show()

# Distribusi Source
plt.figure(figsize=(10, 5))
sns.countplot(data=df, y='source', hue='labels', palette='Set1', order=df['source'].value_counts().index)
plt.title('Distribusi Teks Berdasarkan Sumber Platform & Label')
plt.xlabel('Jumlah Teks')
plt.ylabel('Sumber Platform (Source)')
plt.legend(title='Label (1 = Hate Speech)')
plt.show()

# Panjang Teks
df['text_length'] = df['text'].astype(str).apply(len)

plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='text_length', hue='labels', bins=50, kde=True, palette='Set2')
plt.title('Distribusi Panjang Karakter Teks (Hate Speech vs Non-Hate Speech)')
plt.xlabel('Panjang Karakter')
plt.ylabel('Frekuensi')
plt.xlim(0, 500)
plt.show()

# Top Kata dan Frasa
def get_top_ngrams(corpus, n=1, top_k=15):
    vec = CountVectorizer(ngram_range=(n, n), max_features=10000).fit(corpus)
    bag_of_words = vec.transform(corpus)
    sum_words = bag_of_words.sum(axis=0) 
    words_freq = [(word, sum_words[0, idx]) for word, idx in vec.vocabulary_.items()]
    words_freq = sorted(words_freq, key=lambda x: x[1], reverse=True)
    return words_freq[:top_k]

text_hate = df[df['labels'] == 1]['text'].dropna().astype(str)
text_non_hate = df[df['labels'] == 0]['text'].dropna().astype(str)

print("\n=== TOP 15 KATA (UNIGRAM) - HATE SPEECH (Label 1) ===")
for word, freq in get_top_ngrams(text_hate, n=1, top_k=15):
    print(f"- {word}: {freq}")

print("\n=== TOP 15 KATA (UNIGRAM) - NON-HATE SPEECH (Label 0) ===")
for word, freq in get_top_ngrams(text_non_hate, n=1, top_k=15):
    print(f"- {word}: {freq}")
    
print("\n=== TOP 10 FRASA (BIGRAM) - HATE SPEECH (Label 1) ===")
for phrase, freq in get_top_ngrams(text_hate, n=2, top_k=10):
    print(f"- {phrase}: {freq}")
```

```python
from datasets import load_dataset
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# Load data
hf_token = "..." 

print("Loading dataset...")
ds = load_dataset("manueltonneau/indonesian-hate-speech-superset", token=hf_token)
df = ds['train'].to_pandas()

df['text'] = df['text'].fillna('').astype(str)

# Ekstrak N-grams
def get_top_phrases(corpus, n=2, top_k=15):
    vec = CountVectorizer(ngram_range=(n, n), max_features=10000).fit(corpus)
    bag_of_words = vec.transform(corpus)
    sum_words = bag_of_words.sum(axis=0) 
    words_freq = [(word, sum_words[0, idx]) for word, idx in vec.vocabulary_.items()]
    words_freq = sorted(words_freq, key=lambda x: x[1], reverse=True)
    return words_freq[:top_k]

# Data filter by label
print("Mengekstrak frasa...")
text_hate = df[df['labels'] == 1]['text']
text_non_hate = df[df['labels'] == 0]['text']

print("\n=== TOP 15 FRASA HATE SPEECH (Label 1) ===")
for phrase, freq in get_top_phrases(text_hate, n=2, top_k=15):
    print(f"- {phrase}: {freq}")

print("\n" + "="*45 + "\n")

print("=== TOP 15 FRASA NON-HATE SPEECH (Label 0) ===")
for phrase, freq in get_top_phrases(text_non_hate, n=2, top_k=15):
    print(f"- {phrase}: {freq}")
```
