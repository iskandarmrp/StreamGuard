```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re
from sklearn.feature_extraction.text import CountVectorizer

# Load data
df = pd.read_csv('data/indonesian_abusive_and_hate_speech_twitter.csv', encoding='latin-1')

# Distribusi Label Utama
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.countplot(x='HS', data=df, palette='Set2')
plt.title('Distribusi Hate Speech (HS)')

plt.subplot(1, 2, 2)
sns.countplot(x='Abusive', data=df, palette='Set2')
plt.title('Distribusi Bahasa Abusive')
plt.tight_layout()
plt.show()

# Analisis Target HS
hs_targets = ['HS_Individual', 'HS_Group', 'HS_Religion', 'HS_Race', 'HS_Physical', 'HS_Gender', 'HS_Other']
target_counts = df[hs_targets].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(x=target_counts.values, y=target_counts.index, palette='viridis')
plt.title('Sasaran (Target) Hate Speech')
plt.xlabel('Jumlah Tweet')
plt.tight_layout()
plt.show()

# Tingkat Keparahan HS
hs_severity = ['HS_Weak', 'HS_Moderate', 'HS_Strong']
severity_counts = df[hs_severity].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(x=severity_counts.index, y=severity_counts.values, palette='magma')
plt.title('Tingkat Keparahan Hate Speech')
plt.ylabel('Jumlah Tweet')
plt.tight_layout()
plt.show()

# Cleaning
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'user|\burl\b|rt|xf0|x9f|x98|x82|x8a|xa4|\n|\\n|\\x[0-9a-f]{2}', ' ', text)
    text = re.sub(r'[^a-z0-9 ]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['Clean_Tweet'] = df['Tweet'].apply(clean_text)

def get_top_ngrams(corpus, n=1, top_k=10):
    vec = CountVectorizer(ngram_range=(n, n)).fit(corpus)
    bag_of_words = vec.transform(corpus)
    sum_words = bag_of_words.sum(axis=0) 
    words_freq = [(word, sum_words[0, idx]) for word, idx in vec.vocabulary_.items()]
    words_freq = sorted(words_freq, key=lambda x: x[1], reverse=True)
    return words_freq[:top_k]

# Menampilkan kata teratas
print("Top Kata Hate Speech:")
print(get_top_ngrams(df[df['HS'] == 1]['Clean_Tweet'], n=1, top_k=10))

print("\nTop Kata Non-Hate Speech:")
print(get_top_ngrams(df[df['HS'] == 0]['Clean_Tweet'], n=1, top_k=10))
```

```python
import pandas as pd
import re
from sklearn.feature_extraction.text import CountVectorizer

# Load data
df = pd.read_csv('data/indonesian_abusive_and_hate_speech_twitter.csv', encoding='latin-1')

# Menggabungkan Label
df['is_toxic'] = ((df['HS'] == 1) | (df['Abusive'] == 1)).astype(int)

print("=== DISTRIBUSI LABEL GABUNGAN ===")
print(df['is_toxic'].value_counts())

# Cleaning
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'user|\burl\b|rt|xf0|x9f|x98|x82|x8a|xa4|\n|\\n|\\x[0-9a-f]{2}', ' ', text)
    text = re.sub(r'[^a-z0-9 ]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['Clean_Tweet'] = df['Tweet'].apply(clean_text)

# Ekstraksi Top N-Grams
def get_top_ngrams(corpus, n=1, top_k=15):
    vec = CountVectorizer(ngram_range=(n, n), max_features=10000).fit(corpus)
    bag_of_words = vec.transform(corpus)
    sum_words = bag_of_words.sum(axis=0) 
    words_freq = [(word, sum_words[0, idx]) for word, idx in vec.vocabulary_.items()]
    words_freq = sorted(words_freq, key=lambda x: x[1], reverse=True)
    return words_freq[:top_k]

text_toxic = df[df['is_toxic'] == 1]['Clean_Tweet']
text_clean = df[df['is_toxic'] == 0]['Clean_Tweet']

print("\n=== TOP 15 KATA (UNIGRAM) - HATE/ABUSIVE (Label 1) ===")
for w, f in get_top_ngrams(text_toxic, n=1, top_k=15):
    print(f"- {w}: {f}")

print("\n=== TOP 15 KATA (UNIGRAM) - NON-HATE/ABUSIVE (Label 0) ===")
for w, f in get_top_ngrams(text_clean, n=1, top_k=15):
    print(f"- {w}: {f}")

print("\n=== TOP 15 FRASA (BIGRAM) - HATE/ABUSIVE (Label 1) ===")
for w, f in get_top_ngrams(text_toxic, n=2, top_k=15):
    print(f"- {w}: {f}")

print("\n=== TOP 15 FRASA (BIGRAM) - NON-HATE/ABUSIVE (Label 0) ===")
for w, f in get_top_ngrams(text_clean, n=2, top_k=15):
    print(f"- {w}: {f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re

# Load data
df = pd.read_csv('data/indonesian_abusive_and_hate_speech_twitter.csv', encoding='latin-1')

df['is_toxic'] = ((df['HS'] == 1) | (df['Abusive'] == 1)).astype(int)

# Cleaning
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'user|\burl\b|rt|xf0|x9f|x98|x82|x8a|xa4|\n|\\n|\\x[0-9a-f]{2}', ' ', text)
    text = re.sub(r'[^a-z0-9 ]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['Clean_Tweet'] = df['Tweet'].apply(clean_text)

# Word Count
df['word_count'] = df['Clean_Tweet'].apply(lambda x: len(x.split()))

# Distribusi Label
plt.figure(figsize=(8, 6))
ax = sns.countplot(data=df, x='is_toxic', palette='viridis')
plt.title('Distribusi Label Gabungan (Hate Speech / Abusive)')
plt.xlabel('Label (0: Bersih, 1: Toxic)')
plt.ylabel('Jumlah Tweet')

for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', xytext=(0, 9), textcoords='offset points')

plt.tight_layout()
plt.show()

# Distribusi Panjang Kata
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='word_count', hue='is_toxic', bins=40, kde=True, palette='coolwarm', alpha=0.6)
plt.title('Distribusi Jumlah Kata per Tweet')
plt.xlabel('Jumlah Kata (Word Count)')
plt.ylabel('Frekuensi')
plt.xlim(0, 60) 
plt.tight_layout()
plt.show()
```
