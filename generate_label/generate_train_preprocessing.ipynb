```python
import pandas as pd

df_final = pd.read_csv('../resources/train_final.csv')
df_prep = pd.read_csv('../resources/data_lama_tapi_baru/train_preprocessing_awal.csv')

merge_keys = ['Waktu', 'Username', 'Pesan', 'data_source']

df_final_subset = df_final[merge_keys + ['Label Toxic', 'Label Judi Online']]
df_awal_subset = df_prep[merge_keys + ['Pesan_Clean']]

df_merged = pd.merge(df_awal_subset, df_final_subset, on=merge_keys, how='inner')

df_merged = df_merged[['Waktu', 'Username', 'Pesan', 'Pesan_Clean', 'Label Toxic', 'Label Judi Online', 'data_source']]

df_merged.to_csv('../resources/train_preprocessing_awal.csv', index=False)

print(f"Data berhasil digabungkan. Total baris: {len(df_merged)}")
print(df_merged.head())
```
