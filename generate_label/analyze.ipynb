```python
import pandas as pd

# Load data
file_path = "manual_label/data_chat_youtube_tribun_jogja_sidang_isbat_28_2_2025_cleaned_labeled.csv"

df = pd.read_csv(file_path)

JUMLAH_CONTOH = 10 

# Cek Label Toxic
print("========================================")
print("        ANALISIS LABEL TOXIC")
print("========================================")

print("=== Jumlah Tiap Jenis Label Toxic ===")
print(df['Label Toxic'].value_counts())
print("\n")

print(f"=== {JUMLAH_CONTOH} Contoh Pesan per Label Toxic ===")
unique_toxic_labels = df['Label Toxic'].dropna().unique()

for label in unique_toxic_labels:
    print(f"\n[ Label Toxic: {label} ]")
    examples_toxic = df[df['Label Toxic'] == label].head(JUMLAH_CONTOH)
    
    # Menampilkan Username dan Pesan
    for index, row in examples_toxic.iterrows():
        print(f"- {row['Username']}: {row['Pesan_Clean']}")

print("\n\n")


# Cek Label Judi Online
print("========================================")
print("      ANALISIS LABEL JUDI ONLINE")
print("========================================")

print("=== Jumlah Tiap Jenis Label Judi Online ===")
print(df['Label Judi Online'].value_counts())
print("\n")

print(f"=== {JUMLAH_CONTOH} Contoh Pesan per Label Judi Online ===")
unique_judi_labels = df['Label Judi Online'].dropna().unique()

for label in unique_judi_labels:
    print(f"\n[ Label Judi Online: {label} ]")
    examples_judi = df[df['Label Judi Online'] == label].head(JUMLAH_CONTOH)
    
    # Menampilkan Username dan Pesan
    for index, row in examples_judi.iterrows():
        print(f"- {row['Username']}: {row['Pesan_Clean']}")
```
