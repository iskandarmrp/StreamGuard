```python
import pandas as pd
from sklearn.model_selection import train_test_split
import os

# Test set
test_files = [
    "manual_label/data_chat_youtube_gadgetin_17_4_2024_cleaned_labeled.csv"
]

# Validation set
val_files = [
    "manual_label/data_chat_youtube_dss_music_11_5_2025_cleaned_labeled.csv",
    "manual_label/data_chat_youtube_dss_music_17_9_2021_cleaned_labeled.csv",
    "manual_label/data_chat_youtube_marsha_19_3_2023_cleaned_labeled.csv",
    "manual_label/data_chat_youtube_tribun_jogja_sidang_isbat_28_2_2025_cleaned_labeled.csv"
]

test_df_list = []
val_df_list = []

print("Membaca dan menggabungkan semua file...")
for file in test_files:
    try:
        df_temp = pd.read_csv(file)
        df_temp['data_source'] = file 
        test_df_list.append(df_temp)
    except Exception as e:
        print(f"Gagal membaca {file}: {e}")

for file in val_files:
    try:
        df_temp = pd.read_csv(file)
        df_temp['data_source'] = file 
        val_df_list.append(df_temp)
    except Exception as e:
        print(f"Gagal membaca {file}: {e}")

df_test_all = pd.concat(test_df_list, ignore_index=True)
df_val_all = pd.concat(val_df_list, ignore_index=True)

# Hapus baris kosong
df_test_all = df_test_all.dropna(subset=['Label Toxic', 'Label Judi Online', 'Pesan_Clean', 'Pesan'])
df_val_all = df_val_all.dropna(subset=['Label Toxic', 'Label Judi Online', 'Pesan_Clean', 'Pesan'])

# Hapus duplikat
df_test_all = df_test_all.drop_duplicates(subset=['Pesan'])
df_val_all = df_val_all.drop_duplicates(subset=['Pesan'])
print(f"Total data bersih test (tanpa duplikat pesan): {len(df_test_all)} baris")
print(f"Total data bersih validation (tanpa duplikat pesan): {len(df_val_all)} baris")

# Export Hasil
os.makedirs("../resources", exist_ok=True)

df_test_all.to_csv("../resources/test.csv", index=False)
df_val_all.to_csv("../resources/val.csv", index=False)

print("\nProses sukses! File hasil split telah disimpan di folder '../resources/'")
```

```python
print("\nInformasi Data Test:")
df_test_all.info()
print("\nDistribusi Label Toxic:")
print(df_test_all['Label Toxic'].value_counts())
print("\nDistribusi Label Judi Online:")
print(df_test_all['Label Judi Online'].value_counts())
```

```python
print("\nInformasi Data Validation:")
df_val_all.info()
print("\nDistribusi Label Toxic:")
print(df_val_all['Label Toxic'].value_counts())
print("\nDistribusi Label Judi Online:")
print(df_val_all['Label Judi Online'].value_counts())
```
