```python
import pandas as pd
import re

# Load Data
df = pd.read_csv('data_chat_youtube_bigmo_27_2_2026.csv')

# Keywords
toxic_keywords = ['anjing', 'babi', 'bangsat', 'kontol', 'memek', 'ngentot', 'goblok', 'tolol', 'kampang', 'perek', 'pantek', 'bajingan', 'asu', 'tai', 'ngewe']
judi_keywords = ['gacor', 'slot', 'depo', 'wd', 'maxwin', 'zeus', 'pragmatic', 'judi', 'judol', 'situs', 'link', 'bonus', 'freebet', 'toto', 'togel', 'jp']

df['Pesan'] = df['Pesan'].astype(str)

# Filter Function
def is_toxic(text):
    text_lower = text.lower()
    for kw in toxic_keywords:
        if re.search(r'\b' + kw + r'\b', text_lower):
            return True
    return False

def is_judi(text):
    text_lower = text.lower()
    for kw in judi_keywords:
        if re.search(r'\b' + kw + r'\b', text_lower):
            return True
    return False

df['Toxic'] = df['Pesan'].apply(is_toxic)
df['Judi'] = df['Pesan'].apply(is_judi)

print(f"Total pesan Toxic: {df['Toxic'].sum()}")
print(f"Total pesan Judi: {df['Judi'].sum()}")

print("\n=== 15 Contoh Pesan Toxic ===")
toxic_samples = df[df['Toxic']]['Pesan'].sample(min(15, df['Toxic'].sum())).tolist()
for msg in toxic_samples:
    print(f"- {msg}")

print("\n=== 15 Contoh Pesan Judi / Terkait Judi ===")
judi_samples = df[df['Judi']]['Pesan'].sample(min(15, df['Judi'].sum())).tolist()
for msg in judi_samples:
    print(f"- {msg}")
```
